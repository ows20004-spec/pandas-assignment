```python
titanic = pd.read_csv("data/raw/titanic.csv")      # 저장한 파일 불러오기
print(titanic.shape)                                 # (891, 15)가 나오면 성공
print(titanic.head(3))
```

## 1. 열 선택

```python
#1. fare 열 하나를 선택해 타입을 출력하세요.
print(type(titanic["fare"]))

#2. survived, pclass, sex, age 4개 열을 선택해 타입과 shaape 를 출력하세요.
x = ["survived", "pclass", "sex", "age"]
print(type(titanic[x]))
print(titanic[x].shape)

#3. 1번과 2번의 타입이 다른 이유를 주석 한 줄로 쓰세요.
# 1번은 하나의 열로 나열되는 1차원 데이터 구조이기 때문에 Series이고, 2번은 여러 열을 선택하면서 행과 열로 구성된 2차원 데이터 구조가 되었기 때문에 DataFrame입니다.
```

## 2. loc vs iloc

```python
#1. loc 으로 인덱스 10~15 행의 sex, age 열을 선택해 출력하세요.
print(titanic.loc[10:15, ["sex", "age"]])

#2. iloc 으로 같은 열(sex는 2번째, age는 3번째 위치)을 행 위치 10:15로 선택해 출력하세요.
print(titanic.iloc[10:15,2:4])

#3. 두 결과의 행 수를 각각 출력하고, 왜 다른지 주석으로 쓰세요.
print(len(titanic.loc[10:15, ["sex", "age"]]))
print(len(titanic.iloc[10:15,[2,3]]))
# loc는 행과 열의 '이름'으로 데이터를 선택하기 때문에, 10:15를 입력했을때 이름 그대로 10~15행(6행)이 출력되지만,
# iloc는 행과 열의 위치 기준 슬라이싱이라, 10:15를 입력했을때 10이상 15미만인 5행만 출력됩니다.
```

## 3. iloc 응용

```python
# iloc 으로 마지막 3행의 앞 4개 열만 선택해 출력하세요.(힌트: 음수 위치)
print(titanic.iloc[-3: , :4])
```

## 4. 조건마스크

```python
#1. 나이가 10살 미만인지 검사하는 마스크를 만들어 head(5) 를 출력하세요.
mask = titanic["age"] < 10
print(mask.head(5))

#2. 마스크로 10살 미만 승객 수를 출력하세요.
print(mask.sum())
```

## 5. 복합조건
각각 조건을 만족하는 행 수를 출력하세요

```python
#1. 3등석 그리고 남성 그리고 생존한 승객(&)
mask = (titanic["pclass"] == 3) & (titanic["sex"] == "male") & (titanic["survived"] == 1)
print(mask.sum())

#2. 1등석 또는 요금 100 이상인 승객(|)
mask = (titanic["pclass"] == 1) | (titanic["fare"] >= 100)
print(mask.sum())

#3. 혼자 탑승하지 않은 승객(~, alone 열 사용)
mask = ~titanic["alone"]
print(mask.sum())
```

## 6. isin · between

```python
#1. isin으로 Cherbourg 또는 Queenstown에서 탑승한 승객을 골라 shape를 출력하세요.
mask = titanic["embark_town"].isin(["Cherbourg","Queenstown"])
print(titanic[mask].shape)

#2. between으로 나이 20~30세(양 끝 포함) 승객을 골라 shape와 나이의 min·max를 출력하세요.
mask = titanic["age"].between(20,30)
print(titanic[mask].shape)
x = titanic.loc[mask, "age"]
print(x.min())
print(x.max())


#3. 같은 범위를 양 끝 미포함(inclusive="neither")으로 바꿔 shape를 출력하세요.
mask = titanic["age"].between(20, 30, inclusive = "neither")
print(titanic[mask].shape)
```

## 7. loc[조건,열]

```python
# 12살 이하 그리고 3등석 승객의 age, sex, survived 열만 kids 변수에 담고,
# 행 수와 생존률(servived 평균, 소수 둘째 자리)을 출력하세요.
mask = (titanic["age"] <= 12) & (titanic["pclass"] == 3)
kids = titanic.loc[mask, ["age", "sex","survived"]]

print(len(kids))
print(round(kids["survived"].mean(), 2))
```

## 8. query

```python
#1. 변수 min_fare = 50을 만들고,
#   query로 Cherbourg 탑승 그리고 요금이 min_fare 이상인 승객의 shape를 출력하세요.
min_fare = 50
x = titanic.query("embark_town == 'Cherbourg' & fare >= @min_fare")
print(x.shape)

#2. 5-1번 조건을 query로 다시 써서 행 수가 같은지 확인하세요.
x = titanic.query("pclass == 3 & sex == 'male' & survived == 1")
print(len(x))
# 47행으로 5-1의 행 수와 같습니다.
```

## 9. 정렬

```python
#1. pclass 오름차순, 같은 등급 안에서는 fare 내림차순으로 정렬해
#   pclass, fare, sex, age 열의 상위 3행을 출력하세요.
x = titanic.sort_values(
    by = ["pclass", "fare"],
    ascending=[True, False]
)
print(x[["pclass", "fare", "sex", "age"]].head(3))

#2. 요금 내림차순으로 정렬한 결과를 변수에 담고, iloc[0]과 loc[0]의 fare 값을 각각 출력하세요.
#   두 값이 다른 이유를 주석으로 쓰세요.
x = titanic.sort_values(
    by = "fare",
    ascending=False
)
print(x.iloc[0]["fare"])
print(x.loc[0]["fare"])
# iloc는 정렬된 결과에서 첫번째 위치의 fare값을 불러오고,
# loc는 원래 행의 이름이 '0'이었던 행의 fare값을 가져오기 때문입니다.
```

## 10. 종합 - 20~40세 남성, 1·2등석 vs 3등석
질문 : "20~40세 남성 승객의 생존률은 객실 등급에 따라 얼마나 다를까?"

```python
#1. 남성·나이 20~40세(between)· 1·2등석(isin)조건으로
#   pclass, age, fare, survived 열을 선택하세요.
mask = (titanic["sex"]=="male")&(titanic["age"].between(20,40))&(titanic["pclass"].isin([1,2]))
x = titanic[mask]
male12 = x[["pclass","age","fare","survived"]]
print(male12)

#2. 행 수, 생존률(소수 둘째 자리), 요금 상위 3명을 출력하세요.
print(len(male12))
print(round(male12["survived"].mean(),2))
a = male12.sort_values(by="fare",
                       ascending=False
)
print(a.head(3))

#3. 같은 조건을 query로 써서 행 수를 검증하세요.
x = titanic.query("sex == 'male' & age.between(20,40) & pclass in [1,2]")
print(len(x))
#103행으로 일치


#4. 등급만 3등석으로 바꿔 생존률을 구하고,
#   두 생존률을 비교하는 해석을 마크다운 셀에 2~3문장으로 쓰세요.
mask = (titanic["sex"]=="male")&(titanic["age"].between(20,40))&(titanic["pclass"]==3)
x = titanic[mask]
male3 = x[["pclass","age","fare","survived"]]
print(round(male3["survived"].mean(),2))
```

## 10-4 해석 :

성별(남성)과 연령대(20~40세)를 통제한 상태에서 객실 등급에 따른 생존률을 비교한 결과 1,2등석은 약24%, 3등석은 약15%로 약9%의 차이가 보입니다.

다른 조건이 동일할 때에 객실 등급이 생존률과 관련이 있음을 보여줍니다.

다만, 이 분석은 한 번의 관측 결과에서 나온 두 집단의 단순 비교이기 때문에, 이 차이가 통계적으로 유의한지, 그리고 객실 위치 등 다른 변수의 영향은 없는지 추가적으로 검증이 필요합니다.
